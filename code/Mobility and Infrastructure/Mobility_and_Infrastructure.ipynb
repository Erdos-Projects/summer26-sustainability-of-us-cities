# Category: Mobility & Infrastructure

This category summarizes county-level mobility and infrastructure conditions using a compact PCA-based score based on commuting, broadband access, and population density.

**Raw data location**
- data/social_economic/MobilityData.csv

**Cleaned data with selected variables**
- code/Mobility and Infrastructure/mobility_and_infrastructure.csv

**EDA notebook**
- code/Mobility and Infrastructure/Mobility_and_Infrastructure.ipynb

**Final category score output**
- code/Mobility and Infrastructure/Mobility and Infrastructure Score.csv

## Datasets used and sources

- Mobility dataset from the local social-economic data folder
- Raw source file: data/social_economic/MobilityData.csv

```python
from pathlib import Path
import pandas as pd

root = Path.cwd()
data_path = root / 'data' / 'social_economic' / 'MobilityData.csv'
if not data_path.exists():
    root = root.parent
    data_path = root / 'data' / 'social_economic' / 'MobilityData.csv'

output_dir = root / 'code' / 'Mobility and Infrastructure'
output_dir.mkdir(parents=True, exist_ok=True)

print('Using root:', root)
print('Raw data path exists:', data_path.exists())
```

```python
df = pd.read_csv(data_path)
print(df.head())
print(df.shape)
print(df.columns.tolist())
```

```python
df['FIPS'] = df['state'].astype(int) * 1000 + df['county'].astype(int)
df['County'] = df['county'].astype(str)
selected_vars = ['Long_Commute_Driving_Alone_Raw', 'Broadband_Access_Raw', 'popdensity2010', 'mean_commutetime2000']
clean_df = df[['FIPS', 'County'] + selected_vars].copy()
clean_df['long_commute_shorter'] = 100 - clean_df['Long_Commute_Driving_Alone_Raw']
clean_df['mean_commute_time_lower'] = 100 - clean_df['mean_commutetime2000']
clean_df = clean_df[['FIPS', 'County', 'Broadband_Access_Raw', 'popdensity2010', 'long_commute_shorter', 'mean_commute_time_lower']]
clean_df = clean_df.rename(columns={
    'Broadband_Access_Raw': 'broadband_access_raw',
    'popdensity2010': 'population_density_2010',
    'long_commute_shorter': 'long_commute_shorter',
    'mean_commute_time_lower': 'mean_commute_time_lower',
})
clean_df.to_csv(output_dir / 'mobility_and_infrastructure.csv', index=False)
print(clean_df.head())
```

## Final category score output
