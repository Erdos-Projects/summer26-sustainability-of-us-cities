# Category: Social Capital & Community

This category summarizes county-level social capital and community conditions using a compact PCA-based score based on economic connectedness, volunteering, civic participation, and clustering.

**Raw data location**
- data/social_economic/social_capital_county_economic_connectedness.csv

**Cleaned data with selected variables**
- code/Social Capital and Community/social_capital_and_community.csv

**EDA notebook**
- code/Social Capital and Community/Social_Capital_and_Community.ipynb

**Final category score output**
- code/Social Capital and Community/Social Capital and Community Score.csv

## Datasets used and sources

- Social Capital and Economic Mobility dataset from the local social-economic data folder
- Raw source file: data/social_economic/social_capital_county_economic_connectedness.csv

```python
from pathlib import Path
import pandas as pd

root = Path.cwd()
data_path = root / 'data' / 'social_economic' / 'social_capital_county_economic_connectedness.csv'
if not data_path.exists():
    root = root.parent
    data_path = root / 'data' / 'social_economic' / 'social_capital_county_economic_connectedness.csv'

output_dir = root / 'code' / 'Social Capital and Community'
output_dir.mkdir(parents=True, exist_ok=True)

print('Using root:', root)
print('Raw data path exists:', data_path.exists())
```

```python
df = pd.read_csv(data_path)
selected_vars = [
    'ec_county',
    'child_ec_county',
    'support_ratio_county',
    'volunteering_rate_county',
    'civic_organizations_county',
    'clustering_county',
]
clean_df = df[['county', 'county_name'] + selected_vars].copy()
clean_df = clean_df.rename(columns={'county': 'FIPS', 'county_name': 'County'})
clean_df.to_csv(output_dir / 'social_capital_and_community.csv', index=False)
print(clean_df.head())
```

## Final category score output

The final county-level score is saved as a CSV file in the same folder as the cleaned dataset.
